```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
import pandas as pd
```

```python
train = pd.read_csv("mnist_train.csv")
test = pd.read_csv("mnist_test.csv")
```

```python
x_train = train.iloc[0:, 1:]/255
x_test = test.iloc[0:, 1:]/255
y_train = train.iloc[0:, 0]
y_test = test.iloc[0:, 0]
```

```python
m = keras.Sequential([
    layers.Input((784,)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10,activation='softmax')
])
```

```python
m.compile(
    optimizer=keras.optimizers.SGD(),
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
h = m.fit(x_train, y_train , epochs = 11, batch_size = 128, validation_split = 0.1)
```

```python
print('Test loss accuracy ', m.evaluate(x_test, y_test))
```

```python
for k in ['loss', 'accuracy']:
    plt.plot(h.history[k], label = k)
plt.legend()
plt.show()
```
