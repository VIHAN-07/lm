```python
import pandas as pd
from gensim.models import Word2Vec
from gensim.parsing.preprocessing import preprocess_string
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("twitter_dataset.csv").dropna()[['Text']]
```

```python
tokens = []
for sentences in df['Text']:
    word = preprocess_string(sentences)
    tokens.append(word) 
tokens    
```

```python
cbow = Word2Vec(tokens, sg = 0, window=5, vector_size=50, min_count=2, epochs=10)
```

```python
vectors = []
for token in tokens:
    vect = cbow.wv.get_mean_vector(token)
    vectors.append(vect)
```

```python
query = cbow.wv.get_mean_vector(preprocess_string(input("enter the query")))
df['score'] = cosine_similarity([query], vectors)[0]
```

```python
print(df.nlargest(5, "score")[["Text", "score"]])
```
