```python
from gensim.models import Word2Vec
from gensim.parsing.preprocessing import preprocess_string
import re 
```

```python
file=open("document3.txt", encoding='utf-8')
```

```python
text = file.read().lower()
file.close()
```

```python
sentences = text.split('.')
```

```python
tokens = []
for sentence in sentences:
    s = preprocess_string(sentence)
    tokens.append(s)
```

```python
cbow = Word2Vec(tokens,vector_size = 50 ,window=5, min_count=1, sg = 0, epochs=100)
```

```python
word = preprocess_string("transmission")[0]
print(word)
print("length", len(cbow.wv))
print("word embedding:", cbow.wv[word])
print("word embedding:", cbow.wv.most_similar(word, topn =5))
```
