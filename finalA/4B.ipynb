```python
import tensorflow as tf 
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np 
import pandas as pd 

from sklearn.model_selection import train_test_split

from gensim.models import Word2Vec
```

```python
df = pd.read_csv("twitter_dataset.csv")
df.head()
```

```python
df = df['Text'].str.lower().str.findall("[a-z]+").tolist()
df
```

```python
cbow = Word2Vec(df, vector_size=50, window=5,
                min_count=2, sg=0, epochs=10)
```

```python
word = cbow.wv.index_to_key[0]
print('Sample word:', word)
print('Similar words:', cbow.wv.most_similar(word, topn=5))
```
