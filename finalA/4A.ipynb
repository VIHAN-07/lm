```python
from gensim.models import Word2Vec
import pandas as pd 
from sklearn.preprocessing import LabelBinarizer
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
df = pd.read_csv("IMDB Dataset.csv")
df.head()
```

```python
# df = df.groupby('sentiment').sample(n=1500, random_state = 42)
```

```python
reviews = df['review'].str.lower().str.replace("<br />"," ").str.findall("[a-z]+").tolist()
```

```python
lb = LabelBinarizer()
labels = lb.fit_transform( df['sentiment'])
```

```python
x_train, x_test, y_train, y_test = train_test_split(reviews, labels, test_size= 0.2, random_state= 42) 
```

```python
cbow = Word2Vec(x_train , vector_size=50, window=5, min_count=2, epochs = 5, sg= 0)  
```

```python
import numpy as np
X_train = []
for words in x_train:
    vector = cbow.wv.get_mean_vector(words, pre_normalize=False)
    X_train.append(vector)
X_train = np.array(X_train)  
```

```python
X_test = []
for words in x_test:
    vector = cbow.wv.get_mean_vector(words, pre_normalize=False)
    X_test.append(vector)
X_test = np.array(X_test) 
```

```python
model = keras.Sequential([
    layers.Input(shape=(50,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss = "binary_crossentropy", metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs= 5, batch_size=32)
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("Loss :",loss)
print("test accuracy:",accuracy)
```
