```python

import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# Load and preprocess data
train = pd.read_csv("LP IV Updated Datasets/AutoEncoder/Sign Language MNIST/sign_mnist_train.csv")
test = pd.read_csv("LP IV Updated Datasets/AutoEncoder/Sign Language MNIST/sign_mnist_test.csv")

x_train = train[train.label != 24].drop(columns="label").to_numpy("float32") / 255
x_test = test.drop(columns="label").to_numpy("float32") / 255

# Create Autoencoder
model = Sequential([
    Input(shape=(784,)),
    Dense(32, activation='relu'),
    Dense(784, activation='sigmoid')
])

# Compile
model.compile(
    optimizer='adam',
    loss='mse'
)

# Train
model.fit(x_train, x_train, epochs=10, batch_size=128)

# Detect anomalies
error = np.mean((x_train - model.predict(x_train)) ** 2, axis=1)
threshold = np.percentile(error, 95)

error = np.mean((x_test - model.predict(x_test)) ** 2, axis=1)
anomalies = error > threshold

print("Anomalies:", anomalies.sum())
print("Y Detected:", (anomalies & (test.label == 24)).sum())
```
