```python

import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load data
train = pd.read_csv("LP IV Updated Datasets/FNN and CNN/Fashion MNIST/fashion-mnist_train.csv")
test = pd.read_csv("LP IV Updated Datasets/FNN and CNN/Fashion MNIST/fashion-mnist_test.csv")

x_train = train.iloc[0:, 1:] / 255
y_train = train.iloc[0:, 0]
x_test = test.iloc[0:, 1:] / 255
y_test = test.iloc[0:, 0]

# Create model
model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(10, activation='softmax')
])

# Compile
model.compile(
    optimizer='sgd',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train
history = model.fit(x_train, y_train, epochs=20)

# Evaluate
loss, accuracy = model.evaluate(x_test, y_test)
print("Accuracy:", accuracy * 100, "%")

# Plot
pd.DataFrame(history.history).plot(subplots=True)
```
