```python
from gensim.models import Word2Vec
from gensim.parsing.preprocessing import preprocess_string

from pathlib import Path

dataset_root = Path("LP IV Updated Datasets")
if not dataset_root.is_dir():
    dataset_root = Path("../LP IV Updated Datasets")
print("Dataset folder:", dataset_root.resolve())
```

```python
# Load and inspect document 3
text = (dataset_root / "CBOW/document3.txt").read_text(encoding="utf-8")
print("Document characters:", len(text))
print("Document word count before cleaning:", len(text.split()))
print("Document sample:", text[:200])
```

```python
# Clean and stem sentence segments
sentences = []
for sentence in text.split("."):
    words = preprocess_string(sentence)
    sentences.append(words)
print("Sentence segments:", len(sentences))
print("First sentence tokens:", sentences[0][:20])
```

```python
# Train the Word2Vec CBOW model
cbow = Word2Vec(sentences, min_count=1, sg=0, epochs=100)
print("Vocabulary size:", len(cbow.wv))
print("Word vector size:", cbow.vector_size)
print("Sample vocabulary:", cbow.wv.index_to_key[:10])
```

```python
# Inspect the query embedding and its nearest words
query_word = "transmission"
word = preprocess_string(query_word)[0]
print("Query word:", query_word)
print("Stemmed query:", word)
print("Embedding sample:", cbow.wv[word][:10])
print("Most similar words and cosine scores:")
for similar_word, score in cbow.wv.most_similar(word, topn=5):
    print(similar_word, round(score, 4))
```
