```python
import pandas as pd
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

# Load and preprocess tweets
text = pd.read_csv("LP IV Updated Datasets/CBOW/twitter_dataset.csv").Text.dropna()
sentences = []
for tweet in text:
    words = simple_preprocess(tweet)
    sentences.append(words)

# Train CBOW and find similar words
cbow = Word2Vec(sentences, min_count=1, sg=0, epochs=20)
word = cbow.wv.index_to_key[0]
print(word, cbow.wv.most_similar(word, topn=5))
```
