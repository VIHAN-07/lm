```python

from gensim.models import Word2Vec
from gensim.parsing.preprocessing import preprocess_string

# a. Load and preprocess data
text = open("LP IV Updated Datasets/CBOW/document3.txt", encoding="utf-8").read()
sentences = [preprocess_string(s) for s in text.split(".")]

# b. Train CBOW
cbow = Word2Vec(sentences, min_count=1, sg=0, epochs=100)

# c. Evaluate results
word = preprocess_string("transmission")[0]
print(word, cbow.wv.most_similar(word, topn=5))
```
