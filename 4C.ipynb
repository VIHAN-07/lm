```python
from pathlib import Path
from gensim.models import Word2Vec
from gensim.parsing.preprocessing import preprocess_string

# Clean and stem the document
text = Path("LP IV Updated Datasets/CBOW/document3.txt").read_text(encoding="utf-8")
sentences = []
for sentence in text.split("."):
    words = preprocess_string(sentence)
    sentences.append(words)

# Train CBOW and find similar words
cbow = Word2Vec(sentences, min_count=1, sg=0, epochs=100)
word = preprocess_string("transmission")[0]
print(word, cbow.wv.most_similar(word, topn=5))
```
