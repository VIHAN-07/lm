```python
from pathlib import Path
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

# Load and preprocess the document
text = Path("LP IV Updated Datasets/CBOW/document3.txt").read_text(encoding="utf-8")
sentences = []
for sentence in text.split("."):
    words = simple_preprocess(sentence)
    sentences.append(words)

# Train CBOW and find similar words
cbow = Word2Vec(sentences, min_count=1, sg=0, epochs=100)
print(cbow.wv.most_similar("transmission", topn=5))
```
