```python

import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense
from sklearn.preprocessing import LabelEncoder

# a. Load and preprocess data
train = pd.read_csv("LP IV Updated Datasets/FNN and CNN/Sign Language MNIST/sign_mnist_train.csv", dtype="uint8")
test = pd.read_csv("LP IV Updated Datasets/FNN and CNN/Sign Language MNIST/sign_mnist_test.csv", dtype="uint8")

y_train = train.pop("label")
y_test = test.pop("label")

x_train = train.to_numpy("float32").reshape(-1, 28, 28, 1) / 255
x_test = test.to_numpy("float32").reshape(-1, 28, 28, 1) / 255

encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)

# b. Define CNN architecture
model = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, 3, activation='relu'),
    MaxPooling2D(),
    Flatten(),
    Dense(24, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# c. Train model
model.fit(x_train, y_train, epochs=5, batch_size=128)

# d. Evaluate performance
loss, accuracy = model.evaluate(x_test, y_test)
print("Accuracy:", accuracy * 100, "%")
```
