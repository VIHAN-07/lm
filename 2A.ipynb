```python

import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense

# a. Load and preprocess data
data = pd.read_csv("LP IV Updated Datasets/FNN and CNN/CIFAR10/cifar-10.csv", dtype="uint8")
train, test = train_test_split(data, test_size=0.2)

y_train = train.pop("label")
y_test = test.pop("label")

x_train = train.to_numpy("float32").reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1) / 255
x_test = test.to_numpy("float32").reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1) / 255

# b. Define CNN architecture
model = Sequential([
    Input(shape=(32, 32, 3)),
    Conv2D(32, 3, activation='relu'),
    MaxPooling2D(),
    Flatten(),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# c. Train model
model.fit(x_train, y_train, epochs=5, batch_size=128)

# d. Evaluate performance
loss, accuracy = model.evaluate(x_test, y_test)
print("Accuracy:", accuracy * 100, "%")
```
